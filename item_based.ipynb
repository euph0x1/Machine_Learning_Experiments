```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data= {
    "Users":["U1","U2","U3","U4","U5"],
    "Item1":[5,4,0,2,0],
    "Item2":[3,0,4,0,2],
    "Item3":[4,0,0,3,0],
    "Item4":[0,2,5,0,4]
}
df=pd.DataFrame(data).set_index("Users")
```

```python
i_i_s=pd.DataFrame(cosine_similarity(df.T),
                   index=df.columns,
                   columns=df.columns)
```

```python
target="Item1"
```

```python
score=i_i_s[target]
```

```python
plt.bar(i_i_s.columns,score)
```

```python
rec=score.sort_values(ascending=True)
print(rec)
```
